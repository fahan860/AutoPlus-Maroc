# AUTO+ — Exploration des données fictives (Semaine 1)

Premier notebook d'exploration : génération de données fictives (garages, véhicules) avec Faker
en attendant les vraies données terrain.

```python
import pandas as pd
from faker import Faker

fake = Faker('fr_FR')
Faker.seed(42)

garages = pd.DataFrame([{
    'nom': fake.company(),
    'ville': 'Casablanca',
    'quartier': fake.random_element(['Sidi Maarouf', 'Ain Sebaa', 'Hay Hassani']),
    'note': round(fake.pyfloat(min_value=2.5, max_value=5.0), 1),
} for _ in range(10)])

garages
```

```python
vehicules = pd.DataFrame([{
    'plaque': f"{fake.random_int(1, 99999)}-A-{fake.random_int(1, 99)}",
    'marque': fake.random_element(['Renault', 'Dacia', 'Peugeot', 'Volkswagen', 'Fiat']),
    'annee': fake.random_int(2005, 2025),
} for _ in range(50)])

vehicules.head()
```
